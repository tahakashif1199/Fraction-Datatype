# **Fraction Datatype**

## Abstract Class to inforce some main methods.

```python
from abc import ABC,abstractmethod
class MathDataType(ABC):
  def __init__(self):
    pass

  @abstractmethod
  def __add__(self):
    pass

  @abstractmethod
  def __sub__(self):
    pass

  @abstractmethod
  def __mul__(self):
    pass

  @abstractmethod
  def __truediv__(self):
    pass


```

## Main Class of the Datatype

```python
import math
import random
class Fraction(MathDataType):
  def __init__(self,x,y):
    if y == 0:
      raise TypeError("Denominator cannot be zero.")
    elif type(x) != (int or float):
      raise TypeError("Numerator must be int or float.")
    elif type(y) != (int or float):
      raise TypeError("Denominator must be int or float.")
    else:
      if math.gcd(x,y) != 1:
        self.__num = x/math.gcd(x,y)
        self.__den = y/math.gcd(x,y)
      else:
        self.__num = x
        self.__den = y
        self.__next = 0

  @property
  def num(self):
    return self.__num

  @num.setter
  def num(self,new_num):
    if type(new_num) == ( int or float) and new_num != 0:
      self._Fraction__num = new_num
    else:
      raise ValueError("Numerator cannot be anything but int or float.")

  @property
  def den(self):
    return(self.__den)

  @den.setter
  def den(self,new_num):
    if type(new_num) == ( int or float) and new_num != 0:
      self._Fraction__den = new_num

  def __repr__(self) -> str:
     return f"Fraction({self.__num,self.__den})"

  def __str__(self) -> str:
     return "{}/{}".format(self.__num,self.__den)

# Arithmetic operations
  def __add__(self,other):
    if isinstance(other,Fraction):
     return f"{(self.__num*other.den) + (self.__den*other.__num)}/{(self.__den*other.__den)}"
    elif isinstance(other,int) or isinstance(other,float):
      return f"{(self.__num) + (self.__den*other)}/{(self.__den)}"
    else:
      raise TypeError(f"Addition with {type(other)} is not allowed")


  def __sub__(self,other):
    if isinstance(other,Fraction):
     return f"{(self.__num*other.den) - (self.__den*other.__num)}/{(self.__den*other.__den)}"
    elif isinstance(other,int) or isinstance(other,float):
      return f"{(self.__num) - (self.__den*other)}/{(self.__den)}"
    else:
      raise TypeError(f"Subtration with {type(other)} is not allowed")



  def __mul__(self,other):
    if isinstance(other,Fraction):
     return f"{self.__num*other.__num}/{self.__den*other.__den}"
    elif isinstance(other,int) or isinstance(other,float):
      return f"{self.__num*other}/{self.__den*other}"
    elif other == 0:
      return 0
    elif isinstance(other,list):
      mul_list = []
      for item in other:
        mul_list.append(f"{self.__num*other}/{self.__den*other}")
      return mul_list
    elif isinstance(other,tuple):
      mul_tuple = ()
      for item in other:
        mul_tuple.append(f"{self.__num*other}/{self.__den*other}")
      return mul_tuple
    else:
      raise TypeError(f"Multiplication with {type(other)} is not allowed")

  def __truediv__(self,other):
    if isinstance(other,Fraction):
     return f"{self.__num*other.__den}/{self.__den*other.__num}"
    elif isinstance(other,int) or isinstance(other,float):
      return f"{self.__num*other}/{self.__den}"
    elif other == 0:
      return 0
    else:
      raise ValueError(f"Division is not allowed {type(other)} type object.")

# Equality and comparison
  def __eq__(self,other):
    return True if (self.__num == other.__num) and (self.__den == other.__den) else False

  def __ne__(self,other):
    return True if (self.__num != other.__num) or (self.__den != other.__den) else False

  def __lt__(self,other):
    return True if (self.__num / self.__den) < ( other.__num / other.__den) else False

  def __gt__(self,other):
    return True if (self.__num / self.__den) > ( other.__num / other.__den) else False

  def __le__(self,other):
    return True if (self.__num / self.__den) <= ( other.__num / other.__den) else False

  def __ge__(self,other):
    return True if (self.__num / self.__den) >= ( other.__num / other.__den) else False
# Conversions
  def __int__ (self):
    return int(self.__num/self.__den)

  def __float__(self):
    return self.__num/self.__den

  def __list__(self,*args):
    return [*args]

  def __tuple__(self,*args):
    return tuple(*args)

  def __dict__(self,**kwargs):
    return {**kwargs}

  def __set__(self,*args):
    return set(*args)

# Unary Operators
  def __neg__(self):
    if self.__num< 0 and self.__den < 0:
      return f"{abs(self.__num)}/{abs(self.__den)}"
    elif self.__num < 0 and self.__den > 0:
      return f"{abs(self.__num)}/-{self.__den}"
    elif self.__num > 0 and self.__den < 0:
      return f"-{self.__num}/{abs(self.__den)}"
    elif self.__num > 0 and self.__den > 0:
      return f"-({self.__num}/{self.__den})"

  def __pow__(self,exp):
    return f"{self.__num**exp}/{self.__den**exp}"

  def __abs__(self):
    return f"{abs(self.__num)}/{abs(self.__den)}"

#Methods to make the fration iterable
  def __next__(self):
      if self.__next == 0:
        self.__next += 1
        return self.__num
      elif self.__next == 1:
        return self.__den
      else:
        raise StopIteration("object out of items to iter on.")

  def __iter__(self):
    self.__next = 0
    return self

# Ultility Methods
  def simplify(self):
    ''' Function that simplify the fraction on the basis of gcd
        param self:
        return: simplified fraction'''
    comm_div = math.gcd(self.__num,self.__den)
    return f"{int(self.__num/comm_div)}/{int(self.__den/comm_div)}" if comm_div != 1 else f"{self.__num}/{self.__den}"

  def inverse(self):
    '''Function that inverses the fraction from denominator inplace of numerator and vice versa
      param self:
      return: inversed fraction'''
    return f"{self.__den}/{self.__num}"

  @staticmethod
  def random_frac(nums,min_len=2,max_len=1000,simplify=True):
    '''Function that creats random fractions
      param: nums number of fractions to be created.
      param min_len: minimum value for the selection of the value for the fraction.
      param max_len: maximum value for the selection of the value for the fraction.
      return: a list of randomly created fractions'''
    ran_gen_list = []
    if nums < 0:
      raise ValueError("couldn't generate -ve number of fraction objects. ")
    elif type(nums) != int:
      raise TypeError(f"couldn't generate {type(nums)} number of fraction objects.")
    while nums > 0:
      num = random.randint(min_len,max_len)
      den = random.randint(min_len,max_len)
      ran_gen_list.append(f"{num}/{den}")
      nums -= 1
    return ran_gen_list


```
